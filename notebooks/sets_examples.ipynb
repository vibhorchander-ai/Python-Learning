## Sets — Overview

- A *set* is an unordered collection of unique, hashable elements.
- Duplicate items are automatically removed when creating a set.
- Sets support standard math-like operations: union, intersection, difference, symmetric difference.
- Useful methods: `add`, `remove`, `discard`, `pop`, `clear`, `union`, `intersection`, `difference`, `issubset`, `issuperset`, `isdisjoint`.

Notes:
- Elements must be hashable (immutable) — e.g., strings, numbers, and tuples can be in a set; lists cannot.
- Use `frozenset` when you need an immutable set.

```python
# Creating a set literal: duplicates like repeated 'rock' are removed automatically.
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
# Display the set (order is not guaranteed)
set1
```

```python
# Converting an iterable (like a list) to a set removes duplicates and keeps only unique items.
album_list = ["Michael Jackson", "Thriller", 1982, "00:42:19", \
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
# The resulting set contains the unique elements from `album_list`.
album_set
```

```python
# Use the set constructor with a list; duplicates like repeated 'pop' are removed.
music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul", \
                    "progressive rock", "soft rock", "R&B", "disco"])
music_genres
```

```python
# Define a set and check its type
A = {"Thriller", "Back in Black", "AC/DC"}
A
type(A)  # <class 'set'>
```

```python
# Add an element to the set. If the element already exists, nothing changes.
A.add("NSYNC")
A
```

```python
# Remove an element. `remove` raises KeyError if the element is absent.
A.remove("NSYNC")
A
```

```python
# Membership test — fast average-time complexity O(1)
'AC/DC' in A
```

```python
# Create two sets to demonstrate set algebra (intersection, union, difference)
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
```

```python
# Show both sets (order not guaranteed)
album_set1, album_set2
```

```python
# Intersection: items common to both sets
intersection = album_set1.intersection(album_set2)
intersection  # {'AC/DC', 'Back in Black'}
```

```python
# Difference: items in album_set1 but not in album_set2
difference = album_set1.difference(album_set2)
difference  # {'Thriller'}
```

```python
# Union: all unique items from both sets
union = album_set1.union(album_set2)
union
```

```python
# Check superset/subset relationships
# Is album_set1 a superset of album_set2?
album_set1.issuperset(album_set2)
```

```python
# Is album_set2 a subset of album_set1?
album_set2.issubset(album_set1)
```
