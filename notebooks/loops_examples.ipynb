# Loop Examples

Loops help us repeat actions efficiently. This notebook covers `range()`, `for` loops, `enumerate()`, and `while` loops using short, practical examples.


```python
# range(3) creates a sequence of numbers from 0 to 2.
range(3)
```

```python
# Use a for loop with an index to access each item in a list.
dates = [2020, 2021, 2022]
N = len(dates)
for i in range(N):
    print(dates[i])
```

```python
# A range can start at one value and stop before another value.
for i in range(0,8):
    print(i)
```

```python
# Iterating directly over the list values is simple and readable.
for year in dates:
    print(year)
```

```python
# Update each item in a list while iterating by index.
square = ['red','blue','green','yellow','purple']

for i in range(len(square)):
    print("Before square ", i, "is", square[i])
    square[i] = "white"
    print("After square ", i, "is", square[i])
```

```python
# enumerate() gives both the index and the item value in one loop.
squares = ['red','blue','green','yellow','purple']
for i,color in enumerate(squares):
    print(i, color)
```

```python
# A while loop keeps running until its condition becomes false.
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# This loop continues until the target year is found in the list.
dates = [1982, 1980, 1973, 2000]
i = 0
year = dates[0]
while year != 1973:
    i += 1
    year = dates[i]

print("It took", i, "repitions to get out of the loop")
```
