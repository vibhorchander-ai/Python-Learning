# extend() vs append()
Use `extend()` to add elements from another iterable; use `append()` to add a single element (which can itself be a list).

```python
# Demonstrate extend() vs append(): extend() adds each element, append() adds the object as a single element
L = [ "Michael Jackson", 10.2]
L.extend(['pop', 10])
L
# Append adds a single element (in this case another list) to the end
L.append(['pop1', 11])
L
```

# Mutability
Lists are mutable; assign a new value to an index to change an element in-place.

```python
# Lists are mutable: show how to change an element by index
A = ["disco", 10, 1.2]
print('Before change:', A)
A[0] = 'Hello World!'
print('After change:', A)
```

# Aliasing
Assigning `B = A` creates a reference (alias) to the same list object; mutating one will affect the other.

```python
# Assigning B = A creates a reference (alias) to the same list object. Mutating one affects the other.
A = ["Hello World!", 10, 1.2]
B = A
print('A:', A)
print('B:', B)
```

# Merging lists
Use `extend()` to merge two lists in-place.

```python
# Merge two lists using extend(); A will be modified in-place
A = [1, 'a']
B = [2,1,'d']
A.extend(B)
A
```
