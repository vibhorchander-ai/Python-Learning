```python
# Using with open() is preferred because Python closes the file automatically.
# This means we don't need to call file_data.close() manually.
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    FileContent = file_data.read()
    print(FileContent)
```

```python
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    print(file_data.read(4))
    print(file_data.read(4))
    print(file_data.read(7))
    print(file_data.read(15))
```

```python
# readline() reads one line at a time.
# It can read at most one line per call.
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    print(file_data.readline())
```

```python
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    print(file_data.readline(20))
    print(file_data.read(20))
```

```python
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    i=0
    for line in file_data:
        print(f"Line {i}: {line}")
        i += 1
```

```python
# readlines() reads the file line by line and stores each line as a separate item in a list.
with open("/Users/vibhorchander/Desktop/Python Learning/example1.txt","r") as file_data:
    read = file_data.readlines()
    print(read)
```
