```python
# Dictionary literal: keys can be strings, numbers, or other hashable types (e.g., tuples).
# Keys must be immutable (hashable); values can be any object.
Dict = {'Key1': 1, 'Key2': 2, 'Key3': 3, ('key5',): 5, (0, 1): 6}
# Display the whole dictionary to inspect keys and values
Dict
```

```python
# Access a value by key. Raises KeyError if the key does not exist.
print(Dict['Key1'])  # expected output: 1
```

```python
# Tuple used as a key above — retrieving its corresponding value.
Dict[(0, 1)]  # expected output: 6
```

```python
# Example: mapping album titles to release years.
# Backslash used only for line continuation in this literal for readability.
release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
# Inspect the dictionary
release_year_dict
```

```python
# Retrieve the year for a specific album title
release_year_dict["Thriller"]  # '1982'
```

```python
# View all keys (album titles) as a dynamic view object
release_year_dict.keys()
```

```python
# View all values (release years)
release_year_dict.values()
```

```python
# Add a new key-value pair or update existing one using assignment
release_year_dict["New Album"] = "2024"
# Show updated mapping
release_year_dict
```

```python
# Access an existing entry; raises KeyError if missing
release_year_dict["The Bodyguard"]
```

```python
# Remove a key-value pair using `del`. After deletion the key is absent.
del release_year_dict["The Bodyguard"]
# Show the dictionary after deletion
release_year_dict
```

```python
# Membership test for keys in a dictionary
"Rumours" in release_year_dict  # True if key exists
```
