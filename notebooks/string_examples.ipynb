```python
# Assign a string to Name and print the first 5 characters (slice 0:5)
Name = "The BodyGuard"
print(Name[0:5])
```

```python
# Use extended slicing with a step of 2 to skip every other character
Name[::2]
```

```python
# Demonstrate escape sequence 	 (tab) inside a string
print("The body guard \	 is the best album")
```

```python
# Find the starting index of the substring 'The' (returns -1 if not found)
Name.find('The')
```

```python
# Split the string into words using the default whitespace separator
split_string = Name.split()
split_string
```
