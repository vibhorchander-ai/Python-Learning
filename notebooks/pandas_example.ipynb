```python
# Pandas is a powerful library used for data analysis and manipulation.
# We import it so we can create DataFrames and work with tabular data.
import pandas as pd
```

```python
# Pandas offers a 2-D, mutable-size, heterogeneous table data structure.
# This is called a DataFrame, which stores rows and columns of data.
x = {'Name': ['Rose','John', 'Jane', 'Mary'], 'ID': [1, 2, 3, 4], 'Department': ['Architect Group', 'Software Group', 'Design Team', 'Infrastructure'], 
      'Salary':[100000, 80000, 50000, 60000]}

df = pd.DataFrame(x)

df
```

```python
# Selecting a single column from a DataFrame creates a new DataFrame.
x = df[['ID']]
x
```

```python
# Select multiple columns by listing their names inside a list.
z = df[['ID', 'Salary', 'Department']]
z
```

```python
# A dictionary can be converted into a DataFrame for tabular data analysis.
student_dict = {'Student':['David','Samuel','Terry','Evan'],'Age':[27,24,22,32], 'Country':['UK','Canada','China','USA'], 'Course':['Python','DS','ML','Web Dev'], 'Marks': [85,72,89,76]}
std_df = pd.DataFrame(student_dict)
std_df
```

```python
# Selecting one column keeps the result as a DataFrame when using double brackets.
marks = std_df[['Marks']]
marks
```

```python
# We can also choose multiple columns to view a smaller table.
c = std_df[['Country','Course']]
c
```

```python
# A Series is a one-dimensional labeled array in Pandas.
# It behaves like a single column with an index.
x = std_df['Student']
x
```

```python
# The type() function confirms that a Series is not a DataFrame.
type(x)
```

```python
# .iloc[] selects a value by integer row and column position.
std_df.iloc[0,0]
```

```python
# .loc[] selects a value by row label and column label.
std_df.loc[0,'Marks']
```

```python
# This keeps a copy of the DataFrame for further analysis.
std_df_1 = std_df
# std_df_1 = std_df_1.set_index("Index")
```

```python
# .head() shows the first few rows of the DataFrame.
std_df_1.head()
```

```python
# Access the value in a specific row and column using labels.
std_df_1.loc[0,'Age']
```

```python
# This fetches a value using row and column positions, not labels.
std_df_1.iloc[1,4]
```

```python
# Slice rows and columns to view a subset of the DataFrame.
std_df.iloc[0:2,0:3]
```

```python
# Using .loc with a range of rows and selected columns helps filter data by labels.
std_df.loc[0:2,'Student': 'Marks']
```
