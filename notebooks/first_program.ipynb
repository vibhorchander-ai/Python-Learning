# Common typos
This example intentionally shows a typo to illustrate how NameError or similar errors appear.

```python
# Cell 3: Common typo example
# Notice the misspelling 'frint' instead of 'print'.
# This causes an error because Python does not recognize the function.
frint("Hello, Python!")
```

# Errors stop execution
Demonstrate that an error halts further code in the same cell.

```python
# Cell 4: Demonstrating how Python stops at an error
# The code below prints one line, then fails on the next line.
# Any code after the error will not run.
print("This will be printed")
frint("This will cause an error")
print("This will NOT be printed")
```
