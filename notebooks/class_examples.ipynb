# Class Examples

Classes let us create reusable blueprints for objects. In this notebook, we define a class with attributes and methods, then use that object to build a simple visual example.

```python
# Install the plotting library used for drawing shapes.
import sys

!{sys.executable} -m pip install matplotlib
import matplotlib.pyplot as plt
# Display plots directly in the notebook.
%matplotlib inline
%matplotlib inline
```

```python
# Create a Circle class with a radius and color.
# The constructor initializes the object state for every new instance.
class Circle(object):
    def __init__(self, radius=3,color='blue'):
        self.radius = radius
        self.color = color
    # Add extra size to the circle by increasing its radius.
    def add_radius(self, r):
        self.radius += r
    # Draw the circle on the current plotting area.
    def drawCircle(self):
            plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
            plt.axis('scaled')
            plt.show()
```

```python
# Create a red circle object with radius 10.
RedCircle = Circle(10,'red')
```

```python
# Inspect the object to see its attributes and methods.
dir(RedCircle)
```

```python
# Access the color attribute stored on the object.
RedCircle.color
```

```python
# Read the value of the radius attribute.
RedCircle.radius
```

```python
# Reassign the variable to demonstrate how object references can change.
RedCircle = 1
```

```python
# This call fails because the variable no longer points to a Circle object.
RedCircle.drawCircle()
```

```python
# The value is now an integer, so the radius attribute no longer exists.
RedCircle.radius
```

```python
# Create a fresh circle and then adjust its size with a method.
RedCircle = Circle(10,'red')
RedCircle.add_radius(7)
```

```python
# Draw the updated circle after modifying the radius.
RedCircle.drawCircle()
```

```python
# Create a second circle that uses the default color.
BlueCircle = Circle(100)
```

```python
# Draw the new object to compare it with the earlier circle.
BlueCircle.drawCircle()
```

```python
# Define a second class to show that different object types can be created.
class Rectangle(object):
    def __init__(self,height=3,width=2,color='r'):
        self.height = height
        self.width = width
        self.color = color

    # Draw the rectangle on the same plotting area.
    def drawRectangle(self):
            plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height ,fc=self.color))
            plt.axis('scaled')
            plt.show()
```

```python
# Create a narrow blue rectangle.
SkinnyRectangle = Rectangle(2,3,'blue')
```

```python
# Render the rectangle so the object can be visualized.
SkinnyRectangle.drawRectangle()
```
